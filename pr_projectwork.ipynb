```python
"""
Below code extracts the Olist Brazil e-commerce dataset from kaggle.
There are 9 tables which are extracted and converted into the dataframe.
In case of a current API key fails, then generate a new one from kaggle.com
"""
start_time = time.perf_counter()

# Data collection

# Kaggle credentials
kaggle_username = str(input("Please enter your kaggle user name:"))
kaggle_key = str(input("Please enter your kaggle API key:")) # Generate a new one if the current key fails
link = "olistbr/brazilian-ecommerce" # Replace with the dataset that you want to extract

# Check to ensure the given inputs are accurate before the kaggle APIs pull data
if not all(isinstance(i, str) for i in [kaggle_username,kaggle_key,link]):
  raise TypeError("All configuration inputs must be strings.")

# Invoking the API
path = kagglehub.dataset_download(link)
os.environ["KAGGLE_USERNAME"] = kaggle_username
os.environ["KAGGLE_KEY"] = kaggle_key

# Using a dictionary for faster lookup and fulfills DRY
files = {
    "customers_data": "olist_customers_dataset.csv",
    "location_data": "olist_geolocation_dataset.csv",
    "orders_items_data": "olist_order_items_dataset.csv",
    "orders_payments_data": "olist_order_payments_dataset.csv",
    "orders_reviews_data": "olist_order_reviews_dataset.csv",
    "orders_data": "olist_orders_dataset.csv",
    "products_data": "olist_products_dataset.csv",
    "sellers_data": "olist_sellers_dataset.csv",
    "product_name_translation_data": "product_category_name_translation.csv"
}

datasets = {}

for var_name, filename in files.items():
  file_path = os.path.join(path, filename)
  datasets[var_name] = pd.read_csv(file_path, engine = 'pyarrow')
#The below code helps colab consider the datasets to be global. This ensures the datasets can be accessed from any cell.
globals().update(datasets)

end_time = time.perf_counter()
time_taken = end_time - start_time
print("======================================================Datasets Sucessfully Extracted======================================================")
print("Datasets : Customers || Location || Order Items || Order Payments || Order Reviews || Orders Products || Sellers || Product Category Names")
print(f"Time taken to extract data is: {time_taken:.2f} seconds")
# abhishekhn
# KGAT_3dc11d1511cd353847770c88dc7ac816
```

```python
# ----------------------
# Data cleaning
# ----------------------

# Null check

# customers_data.isna().sum()
# location_data.isnull().sum()
# orders_items_data.isnull().sum()
# orders_payments_data.isnull().sum()
# orders_reviews_data.isnull().sum() - Contains nulls
# orders_data.isnull().sum() - Contains nulls
# products_data.isnull().sum() - Contains nulls
# sellers_data.isnull().sum()
# product_name_translation_data.isnull().sum()

# Filling null values
orders_reviews_data['review_comment_title'] = orders_reviews_data['review_comment_title'].fillna("Sem títulos de avaliação")
orders_reviews_data['review_comment_message'] = orders_reviews_data['review_comment_message'].fillna("Nenhuma avaliação")

products_data['product_category_name'] = products_data['product_category_name'].fillna("unspecified")
products_data['product_name_lenght'] = products_data['product_name_lenght'].fillna(products_data['product_name_lenght'].median())
products_data['product_description_lenght'] = products_data['product_description_lenght'].fillna(products_data['product_description_lenght'].median())
products_data['product_photos_qty'] = products_data['product_photos_qty'].fillna(products_data['product_photos_qty'].median())

products_data['product_weight_g'] = products_data['product_weight_g'].fillna(products_data['product_weight_g'].median())
products_data['product_length_cm'] = products_data['product_length_cm'].fillna(products_data['product_length_cm'].median())
products_data['product_height_cm'] = products_data['product_height_cm'].fillna(products_data['product_height_cm'].median())
products_data['product_width_cm'] = products_data['product_width_cm'].fillna(products_data['product_width_cm'].median())

# Set the columns to the appropriate datatypes
# String based columns
customers_data['customer_id'] = customers_data['customer_id'].astype(str)
customers_data['customer_unique_id'] = customers_data['customer_unique_id'].astype(str)
customers_data['customer_city'] = customers_data['customer_city'].astype(str)
customers_data['customer_state'] = customers_data['customer_state'].astype(str)
location_data['geolocation_city'] = location_data['geolocation_city'].astype(str)
location_data['geolocation_state'] = location_data['geolocation_state'].astype(str)
orders_items_data['order_id'] = orders_items_data['order_id'].astype(str)
orders_items_data['product_id'] = orders_items_data['product_id'].astype(str)
orders_items_data['seller_id'] = orders_items_data['seller_id'].astype(str)
orders_payments_data['order_id'] = orders_payments_data['order_id'].astype(str)
orders_payments_data['payment_type'] = orders_payments_data['payment_type'].astype(str)
orders_reviews_data['review_id'] = orders_reviews_data['review_id'].astype(str)
orders_reviews_data['order_id'] = orders_reviews_data['order_id'].astype(str)
orders_reviews_data['review_comment_title'] = orders_reviews_data['review_comment_title'].astype(str)
orders_reviews_data['review_comment_message'] = orders_reviews_data['review_comment_message'].astype(str)
orders_data['order_id'] = orders_data['order_id'].astype(str)
orders_data['customer_id'] = orders_data['customer_id'].astype(str)
orders_data['order_status'] = orders_data['order_status'].astype(str)
products_data['product_id'] = products_data['product_id'].astype(str)
products_data['product_category_name'] = products_data['product_category_name'].astype(str)
sellers_data['seller_id'] = sellers_data['seller_id'].astype(str)
sellers_data['seller_city'] = sellers_data['seller_city'].astype(str)
sellers_data['seller_state'] = sellers_data['seller_state'].astype(str)
product_name_translation_data['product_category_name'] = product_name_translation_data['product_category_name'].astype(str)
product_name_translation_data['product_category_name_english'] = product_name_translation_data['product_category_name_english'].astype(str)
sellers_data['seller_zip_code_prefix'] = sellers_data['seller_zip_code_prefix'].astype(str)
customers_data['customer_zip_code_prefix'] = customers_data['customer_zip_code_prefix'].astype(str)
location_data['geolocation_zip_code_prefix'] = location_data['geolocation_zip_code_prefix'].astype(str)

# Numerical based columns
location_data['geolocation_lat'] = location_data['geolocation_lat'].astype(float)
location_data['geolocation_lng'] = location_data['geolocation_lng'].astype(float)
orders_items_data['order_item_id'] = orders_items_data['order_item_id'].astype(int)
orders_items_data['price'] = orders_items_data['price'].astype(float)
orders_items_data['freight_value'] = orders_items_data['freight_value'].astype(float)
orders_payments_data['payment_sequential'] = orders_payments_data['payment_sequential'].astype(int)
orders_payments_data['payment_value'] = orders_payments_data['payment_value'].astype(float)
orders_payments_data['payment_installments'] = orders_payments_data['payment_installments'].astype(int)
orders_reviews_data['review_score'] = orders_reviews_data['review_score'].astype(int)
products_data['product_name_lenght'] = products_data['product_name_lenght'].astype(float)
products_data['product_description_lenght'] = products_data['product_description_lenght'].astype(float)
products_data['product_photos_qty'] = products_data['product_photos_qty'].astype(float)
products_data['product_weight_g'] = products_data['product_weight_g'].astype(float)
products_data['product_length_cm'] = products_data['product_length_cm'].astype(float)
products_data['product_height_cm'] = products_data['product_height_cm'].astype(float)
products_data['product_width_cm'] = products_data['product_width_cm'].astype(float)

products_data.rename(columns = {
    'product_name_lenght' : 'product_name_length',
    'product_description_lenght' : 'product_description_length'
}, inplace = True)

# Data-time based columns

orders_items_data['shipping_limit_date'] = pd.to_datetime(orders_items_data['shipping_limit_date'])
orders_reviews_data['review_creation_date'] = pd.to_datetime(orders_reviews_data['review_creation_date'])
orders_reviews_data['review_answer_timestamp'] = pd.to_datetime(orders_reviews_data['review_answer_timestamp'])
orders_data['order_purchase_timestamp'] = pd.to_datetime(orders_data['order_purchase_timestamp'])
orders_data['order_approved_at'] = pd.to_datetime(orders_data['order_approved_at'])
orders_data['order_delivered_carrier_date'] =  pd.to_datetime(orders_data['order_delivered_carrier_date'])
orders_data['order_delivered_customer_date'] =  pd.to_datetime(orders_data['order_delivered_customer_date'])
orders_data['order_estimated_delivery_date'] =  pd.to_datetime(orders_data['order_estimated_delivery_date'])

orders_reviews_data.to_csv("translate_data.csv")

product_name_translation_data['category'] = product_name_translation_data['product_category_name_english'].str.split('_')

# Creating columns
orders_data['delivery_duration_in_days'] = (orders_data['order_estimated_delivery_date'] - orders_data['order_purchase_timestamp']).dt.days
```
